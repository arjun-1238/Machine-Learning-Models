```python
import pandas as pd
df=pd.read_csv("/content/CC GENERAL.csv")
df.columns
display(df['BALANCE'])
```

```python
display(df.value_counts())
display(df.isnull().sum())
```

```python
df=df.drop(["CUST_ID",],axis=1)
df["MINIMUM_PAYMENTS"] = df["MINIMUM_PAYMENTS"].fillna(df["MINIMUM_PAYMENTS"].mean())
df["CREDIT_LIMIT"] = df["CREDIT_LIMIT"].fillna(df["CREDIT_LIMIT"].mean())
display(df.head(5))
```

```python
display(df.value_counts())
display(df.isnull().sum())
```

```python
import numpy as np
numeric_columns=df.select_dtypes(include=['int64','float64']).columns

for col in numeric_columns:
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR=Q3 - Q1

    lower_limit=Q1 - 1.5 * IQR
    upper_limit=Q3 + 1.5 * IQR

    df[col] = np.where(df[col] > upper_limit, upper_limit,
              np.where(df[col] < lower_limit, lower_limit, df[col]))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder,StandardScaler,RobustScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

numeric=df.select_dtypes(include=["int64","float64"]).columns
categorical=df.select_dtypes(include=["object"]).columns
```

```python
num_pipeline = Pipeline(steps=[
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",RobustScaler())
])

cat_pipeline = Pipeline(steps=[
    ("imputer",SimpleImputer(strategy="most_frequent")),
    ("onehot",OneHotEncoder(handle_unknown="ignore",sparse_output=False))
])

preprocessor = ColumnTransformer(transformers=[
    ("num_pipeline",num_pipeline,numeric),
    ("cat_pipeline",cat_pipeline,categorical)
])
```

```python
pipeline=Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("pca",PCA(n_components=5)),
    ("kmeans",KMeans(n_clusters=4,random_state=42,n_init="auto"))
])

display(df.isnull().sum())
df["Cluster"]=pipeline.fit_predict(df)

processed_pca=pipeline.named_steps['pca'].transform(
    pipeline.named_steps['preprocessor'].transform(df)
)
score = silhouette_score(processed_pca, df["Cluster"])
print(f"Silhouette Score:{score:.4f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

proccessed_data = pipeline.named_steps["preprocessor"].transform(df)
inertia_values=[]
k_range=range(1,11)

for k in k_range:
  km=KMeans(n_clusters=k,random_state=42,n_init="auto")
  km.fit(proccessed_data)
  inertia_values.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(k_range,inertia_values,marker="o",linestyle="-")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

```python
pca_2d = PCA(n_components=2)
pca_features_2d=pca_2d.fit_transform(processed_pca)

viz_df = pd.DataFrame(pca_features_2d, columns=['PCA1', 'PCA2'])
viz_df['Cluster'] = df['Cluster']

plt.figure(figsize=(10, 6))
sns.scatterplot(x='PCA1', y='PCA2',hue='Cluster',palette='viridis',data=viz_df,s=50, alpha=0.7)
plt.title("Visualization")
plt.legend(title="Cluster")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
display(df.head(5))-
```
